```python
import pandas as pd
```

```python
#loading the dataset
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns = df.columns.str.replace(' ', '_')
```

```python
df = df.rename(columns={"frequency_of_purchases_days":"purchase_interval_days"})
```

```python
df.isnull().sum()
df['review_rating'] = df.groupby('category')['review_rating'].transform(lambda x:x.fillna(x.median))

```

```python
(df['promo_code_used'] == df['discount_applied']).all()
```

```python
df.drop(columns='promo_code_used')
```

```python
labels = ['young adult','adult','middle aged','senior']
df['age_group'] = pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head()
```

```python



freq_map = {
    "Fortnightly" : 14,
    "Weekly" : 7,
    "Monthly":30,
    "Quarterly":90,
    "Bi-Weekly":14,
    "Annually":365,
    "every 3 months":90
}

df['frequency_of_purchases_days'] = df['frequency_of_purchases'].map(freq_map)


df['frequency_of_purchases_days'].head(10)
```
